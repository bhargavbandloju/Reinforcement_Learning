```python
# Value Iteration in a Deterministic Grid World

# 1. Define Parameters
gamma = 0.9
theta = 0.001

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

normal_reward = -1
goal_reward = 100

# 2. Define Transitions
transitions = {
    'S0': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S0',
        'Right': 'S0'
    },
    'S1': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S1',
        'Right': 'G'
    }
}

# 3. Display MDP Information
print("===== GRID WORLD INFORMATION =====")
print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)
print("Normal Reward:", normal_reward)
print("Goal Reward:", goal_reward)

# 4. Display Transitions
print("\n===== STATE TRANSITIONS =====")

for state in ['S0', 'S1']:
    for action in actions:
        next_state = transitions[state][action]
        reward = goal_reward if next_state == 'G' else normal_reward
        print(state, "--", action, "-->", next_state,
              "| Reward:", reward)

# 5. Initialize State Values
V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

print("\n===== INITIAL STATE VALUES =====")
print(V)

# 6. Implement Value Iteration
iteration = 0

print("\n===== VALUE ITERATION =====")

while True:
    iteration += 1
    new_V = V.copy()

    for state in ['S0', 'S1']:
        q_values = []

        for action in actions:
            next_state = transitions[state][action]

            reward = goal_reward if next_state == 'G' else normal_reward

            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    delta = max(abs(new_V[s] - V[s]) for s in ['S0', 'S1'])

    V = new_V

    print("\nIteration:", iteration)
    print("S0 =", round(V['S0'], 4))
    print("S1 =", round(V['S1'], 4))
    print("G =", V['G'])
    print("Delta =", round(delta, 6))

    if delta < theta:
        break

# 7. Display Final State Values
print("\n===== FINAL OPTIMAL STATE VALUES =====")

for state in states:
    print(state, "=", round(V[state], 4))

# 8. Calculate Action Values
print("\n===== ACTION VALUE FUNCTION Q(s,a) =====")

Q = {}

for state in ['S0', 'S1']:
    Q[state] = {}

    for action in actions:
        next_state = transitions[state][action]

        reward = goal_reward if next_state == 'G' else normal_reward

        q = reward + gamma * V[next_state]

        Q[state][action] = q

        print(state, action, "=", round(q, 4))

# 9. Extract Optimal Policy
print("\n===== OPTIMAL POLICY =====")

for state in ['S0', 'S1']:
    max_value = max(Q[state].values())

    best_actions = [
        action for action, value in Q[state].items()
        if abs(value - max_value) < 1e-9
    ]

    print(state, "->", best_actions)

# 10. Verify Bellman Optimality
print("\n===== BELLMAN OPTIMALITY VERIFICATION =====")

for state in ['S0', 'S1']:
    max_q = max(Q[state].values())

    print("\nState:", state)
    print("V*(s) =", round(V[state], 4))
    print("Max Q*(s,a) =", round(max_q, 4))

    if abs(V[state] - max_q) < 1e-9:
        print("Verification: PASSED")
    else:
        print("Verification: FAILED")

# 11. Final Result
print("\n===== FINAL RESULT =====")
print("V*(S0) =", round(V['S0'], 4))
print("V*(S1) =", round(V['S1'], 4))
print("V*(G) =", V['G'])
print("Optimal Action at S0: Down")
print("Optimal Action at S1: Right")
print("Optimal Path: S0 -> S1 -> G")

print("\nValue Iteration Completed Successfully!")
```
