```python
# Bellman Optimality and Value Iteration in a Deterministic Grid World

# Discount factor
gamma = 0.9

# Convergence threshold
theta = 0.001

# States
states = ['S0', 'S1', 'G']

# Actions
actions = ['Up', 'Down', 'Left', 'Right']

# Rewards
normal_reward = -1
wall_reward = -1
goal_reward = 10

# Transition mapping
# Hitting a boundary or wall keeps the agent in the same state.

transitions = {
    'S0': {
        'Up': ('S0', normal_reward),
        'Down': ('S1', normal_reward),
        'Left': ('S0', normal_reward),
        'Right': ('S0', wall_reward)
    },

    'S1': {
        'Up': ('S0', normal_reward),
        'Down': ('S1', normal_reward),
        'Left': ('S1', normal_reward),
        'Right': ('G', goal_reward)
    }
}

print("STATES:", states)
print("ACTIONS:", actions)
print("DISCOUNT FACTOR:", gamma)
print("NORMAL REWARD:", normal_reward)
print("WALL REWARD:", wall_reward)
print("GOAL REWARD:", goal_reward)
```

```python
print("STATE TRANSITION DETAILS")

for state in ['S0', 'S1']:
    for action in actions:
        next_state, reward = transitions[state][action]

        print("Current State:", state,
              "| Action:", action,
              "| Next State:", next_state,
              "| Reward:", reward)
```

```python
# Initialize all state values to zero

V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

print("INITIAL STATE VALUES")

for state, value in V.items():
    print(state, "=", value)
```

```python
iteration = 0

print("VALUE ITERATION PROCESS")

while True:

    iteration += 1

    # Create a copy for synchronous updates
    new_V = V.copy()

    # Update non-terminal states
    for state in ['S0', 'S1']:

        q_values = []

        for action in actions:

            next_state, reward = transitions[state][action]

            q = reward + gamma * V[next_state]

            q_values.append(q)

        new_V[state] = max(q_values)

    # Calculate maximum change
    delta = max(abs(new_V[s] - V[s]) for s in ['S0', 'S1'])

    V = new_V

    print("\nIteration:", iteration)
    print("S0 =", round(V['S0'], 4))
    print("S1 =", round(V['S1'], 4))
    print("G  =", V['G'])
    print("Delta =", round(delta, 6))

    # Check convergence
    if delta < theta:
        break

print("\nVALUE ITERATION CONVERGED!")
```

```python
print("FINAL OPTIMAL STATE VALUES")

for state, value in V.items():
    print(state, "=", round(value, 4))
```

```python
print("ACTION VALUE FUNCTION Q(s,a)")

Q = {}

for state in ['S0', 'S1']:

    Q[state] = {}

    for action in actions:

        next_state, reward = transitions[state][action]

        q = reward + gamma * V[next_state]

        Q[state][action] = q

        print(state, action, "=", round(q, 4))
```

```python
print("OPTIMAL POLICY")

for state in ['S0', 'S1']:

    max_value = max(Q[state].values())

    optimal_actions = [
        action for action, value in Q[state].items()
        if abs(value - max_value) < 1e-9
    ]

    print(state, "->", optimal_actions)
```

```python
print("BELLMAN OPTIMALITY VERIFICATION")

for state in ['S0', 'S1']:

    max_q = max(Q[state].values())

    print("\nState:", state)
    print("V*(s) =", round(V[state], 4))
    print("Max Q*(s,a) =", round(max_q, 4))

    if abs(V[state] - max_q) < 1e-9:
        print("Verification: PASSED")
    else:
        print("Verification: FAILED")
```

```python
print("EXPLANATION OF RESULTS")

print("1. The optimal action in S0 is Down.")
print("2. The optimal action in S1 is Right.")
print("3. Reaching the goal provides a reward of +10.")
print("4. Moving to a normal cell gives a reward of -1.")
print("5. Hitting a wall gives a reward of -1.")
print("6. The discount factor gamma = 0.9 reduces future rewards.")
print("7. Value Iteration finds the optimal state values.")
print("8. The optimal policy guides the agent towards the goal.")
```
